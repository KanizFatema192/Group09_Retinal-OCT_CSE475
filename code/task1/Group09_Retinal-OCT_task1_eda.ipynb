```python
# Cell 4
# DATASET CLEANING & VERIFICATION REPORT (OCTDL Dataset)

# Dataset root directory
DATASET_DIR = "/kaggle/input/datasets/orvile/octdl-optical-coherence-tomography-dataset"

# OCTDL class names (must match folder names exactly)
classes = ['AMD', 'DME', 'ERM', 'NO', 'RAO', 'RVO', 'VID']
class_to_idx = {cls: idx for idx, cls in enumerate(classes)}

seen_hashes = set()
clean_filepaths = []
clean_labels = []

total_raw_scanned = 0
corrupt_count = 0
duplicate_count = 0

# Scan dataset
for root, _, files in os.walk(DATASET_DIR):

    folder_name = os.path.basename(root)

    if folder_name not in class_to_idx:
        continue

    label = class_to_idx[folder_name]

    for file in files:

        if not file.lower().endswith(
            (".png", ".jpg", ".jpeg", ".bmp", ".tif", ".tiff")
        ):
            continue

        total_raw_scanned += 1

        filepath = os.path.join(root, file)

        # Verify image
        try:
            with Image.open(filepath) as img:
                img.verify()
        except Exception:
            corrupt_count += 1
            continue

        # SHA256 hash
        hasher = hashlib.sha256()

        with open(filepath, "rb") as f:
            while True:
                chunk = f.read(8192)
                if not chunk:
                    break
                hasher.update(chunk)

        file_hash = hasher.hexdigest()

        if file_hash in seen_hashes:
            duplicate_count += 1
            continue

        seen_hashes.add(file_hash)
        clean_filepaths.append(filepath)
        clean_labels.append(label)

# Summary

print("DATASET CLEANING & VERIFICATION REPORT:")

print(f"Total Raw Files Scanned      : {total_raw_scanned:,}")
print(f"Corrupt Images Found         : {corrupt_count:,}")
print(f"Duplicate Images Removed     : {duplicate_count:,}")
print(f"Total Unique Images Kept     : {len(clean_filepaths):,}")

# Class Distribution

total_unique = len(clean_filepaths)

class_counts = {cls: 0 for cls in classes}

for label in clean_labels:
    class_counts[classes[label]] += 1

print("\n")

print("CLASS DISTRIBUTION")


print(f"{'Class':<10}{'Images':>12}{'Percentage':>15}")

for cls in classes:

    count = class_counts[cls]

    percent = (count / total_unique) * 100 if total_unique else 0

    print(f"{cls:<10}{count:>12,}{percent:>14.2f}%")


print(f"{'Total':<10}{total_unique:>12,}{100:>14.2f}%")
```

```python
# Cell 7
# Task 1 — Section B: Summary Statistics & Outliers
np.random.seed(42)

sample_size = min(1000, len(clean_filepaths))
sample_index = np.random.choice(len(clean_filepaths), sample_size, replace=False)

mean_values = []
median_values = []
std_values = []
min_values = []
max_values = []
image_classes = []

# Calculate statistics for each sampled image

for index in sample_index:

    image_path = clean_filepaths[index]
    label = clean_labels[index]

    with Image.open(image_path) as img:

        image = np.array(img.convert("L"), dtype=np.float32)

    mean_values.append(np.mean(image))
    median_values.append(np.median(image))
    std_values.append(np.std(image))
    min_values.append(np.min(image))
    max_values.append(np.max(image))
    image_classes.append(classes[label])
    
# Create DataFrame

df_stats = pd.DataFrame({

    "Class": image_classes,
    "Mean": mean_values,
    "Median": median_values,
    "Std": std_values,
    "Min": min_values,
    "Max": max_values

})

# Overall statistics

q1 = np.percentile(df_stats["Mean"], 25)
q3 = np.percentile(df_stats["Mean"], 75)

iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

outliers = df_stats[
    (df_stats["Mean"] < lower_limit) |
    (df_stats["Mean"] > upper_limit)
]

section_b_table = pd.DataFrame({

    "Statistic": [
        "Mean",
        "Median",
        "Standard Deviation",
        "Minimum",
        "Maximum",
        "First Quartile (Q1)",
        "Third Quartile (Q3)",
        "Interquartile Range (IQR)",
        "Skewness",
        "Kurtosis",
        "Outliers"
    ],

    "Value": [

        f"{df_stats['Mean'].mean():.4f}",
        f"{df_stats['Mean'].median():.4f}",
        f"{df_stats['Mean'].std():.4f}",
        f"{df_stats['Mean'].min():.4f}",
        f"{df_stats['Mean'].max():.4f}",
        f"{q1:.4f}",
        f"{q3:.4f}",
        f"{iqr:.4f}",
        f"{skew(df_stats['Mean']):.4f}",
        f"{kurtosis(df_stats['Mean']):.4f}",
        f"{len(outliers)} ({len(outliers)/sample_size*100:.2f}%)"

    ]

})

print("TASK 1 : SECTION B - OVERALL STATISTICS")
display(section_b_table)

# Class-wise statistics

class_summary = df_stats.groupby("Class").agg(

    Mean_Intensity=("Mean", "mean"),
    Median_Intensity=("Median", "median"),
    Std_Deviation=("Std", "mean"),
    Minimum=("Min", "min"),
    Maximum=("Max", "max"),
    Skewness=("Mean", lambda x: skew(x)),
    Kurtosis=("Mean", lambda x: kurtosis(x))

).reset_index()

print("TASK 1 : SECTION B - CLASS-WISE STATISTICS")
display(class_summary)

# Box Plot 
fig = px.box( 
    df_stats, x="Class", 
    y="Mean", 
    color="Class", 
    points="outliers", 
    title="Mean Pixel Intensity Distribution by Class and Outlier Analysis") 
fig.update_layout( 
    template="plotly_white",
    height=500) 
fig.show()
```

```python
# Cell 8
# Task 1 — Section C: Data Quality & Modeling Impact Analysis
missing_count = 0

for img_path in clean_filepaths:
    if not os.path.exists(img_path):
        missing_count += 1

total_images = len(clean_filepaths)

missing_percent = (missing_count / total_images) * 100
duplicate_percent = (duplicate_count / total_images) * 100

# Pixel Feature Analysis

np.random.seed(42)

sample_size = min(500, total_images)
sample_index = np.random.choice(total_images, sample_size, replace=False)

pixel_data = []

for index in sample_index:

    image_path = clean_filepaths[index]

    with Image.open(image_path) as img:

        image = img.convert("L").resize((32, 32))
        pixel_data.append(np.array(image).flatten())

pixel_data = np.array(pixel_data)

# Near-constant features
pixel_variance = np.var(pixel_data, axis=0)

near_constant = np.sum(pixel_variance < 1.0)
near_constant_percent = (near_constant / pixel_data.shape[1]) * 100

# Highly correlated features
pixel_df = pd.DataFrame(pixel_data)

corr = pixel_df.corr().abs()

upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))

high_corr = sum((upper > 0.90).any())

high_corr_percent = (high_corr / pixel_data.shape[1]) * 100

# Outlier Detection

mean_intensity = pixel_data.mean(axis=1)

q1 = np.percentile(mean_intensity, 25)
q3 = np.percentile(mean_intensity, 75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = np.sum((mean_intensity < lower) | (mean_intensity > upper))

outlier_percent = (outliers / sample_size) * 100

# Summary Table

section_c_summary = pd.DataFrame({

    "Data Quality Aspect": [

        "Missing Images",
        "Duplicate Images",
        "Near-Constant Features",
        "Highly Correlated Features (r > 0.90)",
        "Outliers (Tukey 1.5 × IQR)"

    ],

    "Result": [

        f"{missing_count} ({missing_percent:.2f}%)",
        f"{duplicate_count} ({duplicate_percent:.2f}%)",
        f"{near_constant} / {pixel_data.shape[1]} ({near_constant_percent:.2f}%)",
        f"{high_corr} / {pixel_data.shape[1]} ({high_corr_percent:.2f}%)",
        f"{outliers} / {sample_size} ({outlier_percent:.2f}%)"

    ]

})

print("TASK 1 : SECTION C - DATA QUALITY ANALYSIS")
display(section_c_summary)
```

```python
# Cell 18
# Task 1 — Section F: Correlation & Redundancy Analysis
np.random.seed(42)

GRID_SIZE = (16, 16)
CORR_THRESHOLD = 0.

# Select images for analysis

sample_size = min(600, len(clean_filepaths))
sample_index = np.random.choice(len(clean_filepaths), sample_size, replace=False)

spatial_features = []
image_features = []

for index in sample_index:

    image_path = clean_filepaths[index]
    label = clean_labels[index]

    with Image.open(image_path) as img:

        gray = img.convert("L")
        image = np.array(gray, dtype=np.float32)

        # 16 × 16 spatial features
        small_image = gray.resize(GRID_SIZE)
        spatial_features.append(np.array(small_image).flatten())

        # Global image statistics
        image_features.append({

            "Class": classes[label],
            "Mean": image.mean(),
            "Median": np.median(image),
            "Std": image.std(),
            "Min": image.min(),
            "Max": image.max(),
            "Q1": np.percentile(image, 25),
            "Q3": np.percentile(image, 75)

        })

# Create DataFrames

df_spatial = pd.DataFrame(
    spatial_features,
    columns=[f"Region_{i+1}" for i in range(256)]
)

df_stats = pd.DataFrame(image_features)

# Correlation Heatmap

feature_data = df_stats.drop(columns="Class").copy()

feature_data["Region_1"] = df_spatial["Region_1"]
feature_data["Region_64"] = df_spatial["Region_64"]
feature_data["Region_128"] = df_spatial["Region_128"]
feature_data["Region_192"] = df_spatial["Region_192"]
feature_data["Region_256"] = df_spatial["Region_256"]

corr_matrix = feature_data.corr()

fig = px.imshow(

    corr_matrix,

    text_auto=".2f",

    color_continuous_scale="RdBu_r",

    zmin=-1,
    zmax=1,

    title="Feature Correlation Matrix"

)

fig.update_layout(template="plotly_white")

fig.show()

# Redundant Features

corr = df_spatial.corr().abs()

upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))

redundant = []

for column in upper.columns:

    rows = upper.index[upper[column] > CORR_THRESHOLD]

    for row in rows:

        redundant.append({

            "Feature 1": row,
            "Feature 2": column,
            "Correlation": round(corr.loc[row, column], 4)

        })

redundant_df = pd.DataFrame(redundant)

print("TASK 1 : SECTION F - REDUNDANT FEATURES")

if len(redundant_df) == 0:

    print("No redundant feature pairs found.")

else:

    display(redundant_df.head(15))

    print(f"\nTotal Redundant Feature Pairs : {len(redundant_df)}")
```

```python
# Cell 19
# Task 1 — Section G: Dimensionality Reduction (PCA / t-SNE / UMAP)
np.random.seed(42)

IMAGE_SIZE = (64, 64)

sample_size = min(1200, len(clean_filepaths))

sample_index = np.random.choice(
    len(clean_filepaths),
    sample_size,
    replace=False
)

X = []
y = []

for index in sample_index:

    image = Image.open(clean_filepaths[index]).convert("L")

    image = image.resize(IMAGE_SIZE)

    image = np.array(image, dtype=np.float32) / 255.0

    X.append(image.flatten())

    y.append(classes[clean_labels[index]])

X = np.array(X)
```

```python
# Cell 20
# Task 1 — Section G: Dimensionality Reduction (PCA / t-SNE / UMAP)
# 2D PCA
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X)

variance = pca.explained_variance_ratio_.sum() * 100

df_pca = pd.DataFrame(X_pca, columns=["PC1", "PC2"])

df_pca["Class"] = y

fig = px.scatter(df_pca, x="PC1", y="PC2", color="Class",

title=f"2D PCA (Explained Variance = {variance:.2f}%)" )

fig.update_layout(template="plotly_white")
fig.show()

# 3D PCA
pca = PCA(n_components=3)

X_pca = pca.fit_transform(X)

df = pd.DataFrame(X_pca, columns=["PC1","PC2","PC3"])

df["Class"] = y

fig = px.scatter_3d(df, x="PC1", y="PC2", z="PC3", color="Class",
title=f"3D PCA (Explained Variance = {variance:.2f}%)" )

fig.update_layout(template="plotly_white")
fig.show()
```
